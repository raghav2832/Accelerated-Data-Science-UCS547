1.Reverse the NumPy array: arr = np.array([1, 2, 3, 6, 4, 5]) )

```python
import numpy as np

arr = np.array([7, 11, 4, 9, 2, 15])

rev_arr = arr[::-1]

print("Original Array:", arr)
print("Reversed Array:", rev_arr)
```

3.Find the most frequent value and their indice(s) in the following arrays:
i.x = np.array([1,2,3,4,5,1,2,1,1,1])
ii.y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])

```python
import numpy as np

x = np.array([3, 7, 3, 9, 7, 3, 3, 5, 7])

values, counts = np.unique(x, return_counts=True)
max_count = counts.max()

most_frequent = values[counts == max_count]

print("Array:", x)
print("Most Frequent Value(s):", most_frequent)

for val in most_frequent:
    indices = np.where(x == val)
    print(f"Indices of {val}:", indices[0])
```

```python
import numpy as np

y = np.array([4, 6, 4, 8, 6, 4, 9, 6, 6])

values, counts = np.unique(y, return_counts=True)
max_count = counts.max()

most_frequent = values[counts == max_count]

print("Array:", y)
print("Most Frequent Value(s):", most_frequent)

for val in most_frequent:
    indices = np.where(y == val)
    print(f"Indices of {val}:", indices[0])
```

4.For the array gfg = np.matrix('[4, 1, 9; 12, 3, 1; 4, 5, 6]'),
find i. Sum of all elements
ii.
iii.
Sum of all element’s row-wise
Sum of all element’s column-wise

```python
import numpy as np

mat = np.matrix('[7, 2, 5; 1, 9, 3; 4, 6, 8]')

total_sum = mat.sum()
row_sum = mat.sum(axis=1)
column_sum = mat.sum(axis=0)

print("Matrix:\n", mat)
print("Sum of all elements:", total_sum)
print("Row-wise sum:\n", row_sum)
print("Column-wise sum:\n", column_sum)
```
